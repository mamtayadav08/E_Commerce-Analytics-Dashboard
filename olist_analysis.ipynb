```python
df = pd.merge(
    orders,
    customers,
    on="customer_id",
    how="left"
)
```

```python
df = pd.merge(
    df,
    payments,
    on="order_id",
    how="left"
)
```

```python
total_revenue = df["payment_value"].sum()

total_orders = df["order_id"].nunique()

aov = total_revenue / total_orders

print(round(aov, 2))
```

The Average Order Value (AOV) is approximately 160.99, indicating moderate customer spending per transaction across the platform.

```python
repeat_customers = df.groupby("customer_unique_id")["order_id"].nunique()
```

```python
repeat_rate = (repeat_customers > 1).mean() * 100

print(round(repeat_rate, 2))
```

The Repeat Customer Rate was approximately 3.12%, indicating that the majority of customers made only one purchase, highlighting potential opportunities to improve customer retention strategies.

```python
purchase_frequency = df["order_id"].nunique() / df["customer_unique_id"].nunique()

print(round(purchase_frequency, 2))
```

Purchase Frequency was approximately 1.03 orders per customer, suggesting that most customers placed only a single order during the observed period.

```python
df["order_purchase_timestamp"] = pd.to_datetime(df["order_purchase_timestamp"])
```

```python
monthly_revenue = df.groupby(
    df["order_purchase_timestamp"].dt.to_period("M")
)["payment_value"].sum()
```

```python
print(monthly_revenue)
```

```python
import matplotlib.pyplot as plt

monthly_revenue.plot(figsize=(12,5))

plt.title("Monthly Revenue Trend")

plt.xlabel("Month")

plt.ylabel("Revenue")

plt.show()
```

```python
df["orders_hour"] = pd.to_datetime(
    df["order_purchase_timestamp"]
).dt.hour
```

```python
hourly_orders = df.groupby("orders_hour")["order_id"].count()
```

```python
print(hourly_orders)
```

```python
import matplotlib.pyplot as plt

hourly_orders.plot(figsize=(10,5))

plt.title("Orders by Hour")

plt.xlabel("Hour of Day")

plt.ylabel("Number of Orders")

plt.show()
```

Customer purchasing activity was highest between 10 AM and 9 PM, with peak order volume around 4 PM. Businesses can optimize promotional campaigns and notifications during these high-engagement hours.

```python
import pandas as pd

orders = pd.read_csv("../Data/raw/olist_orders_dataset.csv")

customers = pd.read_csv("../Data/raw/olist_customers_dataset.csv")

payments = pd.read_csv("../Data/raw/olist_order_payments_dataset.csv")

items = pd.read_csv("../Data/raw/olist_order_items_dataset.csv")

products_scraped = pd.read_csv("../Data/raw/ecommerce_products.csv")
```

```python
category_price = products_scraped.groupby("Category")["Price"].mean()

category_price = category_price.sort_values(ascending=False)

print(category_price)
```

```python
import matplotlib.pyplot as plt

category_price.plot(kind="bar", figsize=(10,5))

plt.title("Average Price by Category")

plt.xlabel("Category")

plt.ylabel("Average Price")

plt.show()
```

```python
payment_revenue = df.groupby("payment_type")["payment_value"].sum()

payment_revenue = payment_revenue.sort_values(ascending=False)

print(payment_revenue)
```

```python
import matplotlib.pyplot as plt

payment_revenue.plot(kind="bar", figsize=(8,5))

plt.title("Revenue by Payment Type")

plt.xlabel("Payment Type")

plt.ylabel("Revenue")

plt.show()
```

```python
monthly_orders = df.groupby(
    df["order_purchase_timestamp"].dt.to_period("M")
)["order_id"].nunique()

print(monthly_orders)
```

```python
import matplotlib.pyplot as plt

monthly_orders.plot(
    kind="line",
    marker="o",
    figsize=(12,5)
)

plt.title("Monthly Order Volume Trend")
plt.xlabel("Month")
plt.ylabel("Number of Orders")

plt.grid(True)

plt.show()
```

```python
df["order_purchase_timestamp"] = pd.to_datetime(
    df["order_purchase_timestamp"]
)

df["order_delivered_customer_date"] = pd.to_datetime(
    df["order_delivered_customer_date"]
)

delivery_days = (
    df["order_delivered_customer_date"]
    - df["order_purchase_timestamp"]
).dt.days

print(round(delivery_days.mean(), 2))
```

```python
import matplotlib.pyplot as plt

delivery_days.hist(bins=30)

plt.title("Delivery Time Distribution")

plt.xlabel("Delivery Days")

plt.ylabel("Number of Orders")

plt.show()
```

```python
category_revenue = products_scraped.groupby(
    "Category"
)["Price"].sum()

category_revenue = category_revenue.sort_values(
    ascending=False
)

print(category_revenue)
```
