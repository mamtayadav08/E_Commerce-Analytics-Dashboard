```python
import requests
import pandas as pd
```

```python
url = "https://fakestoreapi.com/products"
```

```python
response = requests.get(url)
```

```python
print(response.status_code)
```

```python
data = response.json()
```

```python
print(data[0])
```

```python
products = []
```

```python
for item in data:
    print (item)
```

```python
for item in data:

    title = item["title"]

    print(title)

    print("-----")
```

```python
for item in data:

    title = item["title"]

    price = item["price"]

    category = item["category"]

    print(title)

    print(price)

    print(category)

    print("------")
```

```python
for item in data:

    rating = item["rating"]["rate"]

    print(rating)
```

```python
for item in data:

    title = item["title"]

    price = item["price"]

    category = item["category"]

    rating = item["rating"]["rate"]

    products.append({
        "Title": title,
        "Price": price,
        "Category": category,
        "Rating": rating
    })
```

```python
print(products)
```

```python
df = pd.DataFrame(products)
```

```python
print(df.head())
```

```python
df.to_csv("ecommerce_products.csv", index=False)
```
